```python
import pandas as pd

df = pd.read_csv("/content/superstore_sales (1).csv", encoding="latin1")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df.head())
```

```python
print("Missing values in each column:")
display(df.isnull().sum())

print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Rows missing Order Date:")
display(df[df["Order Date"].isna()].head(10))
```

```python
df_clean = df.dropna(
    subset=["Order Date", "Sales", "Profit"]
).copy()

print("Original rows:", len(df))
print("Valid sales rows:", len(df_clean))
print("Rows excluded:", len(df) - len(df_clean))
```

```python
print("Missing values in valid sales data:")
display(df_clean.isnull().sum())

print("Total missing values:", df_clean.isnull().sum().sum())
```

```python
df_clean["Postal Code"] = df_clean["Postal Code"].fillna(5401)

df_clean["Postal Code"] = (
    df_clean["Postal Code"]
    .astype(int)
    .astype(str)
    .str.zfill(5)
)

print("Missing Postal Codes:", df_clean["Postal Code"].isnull().sum())
print("Postal Code cleaning completed!")
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())

print("Total valid sales rows:", len(df_clean))
```

```python
df_clean["Order Date"] = pd.to_datetime(
    df_clean["Order Date"],
    errors="coerce"
)

df_clean["Ship Date"] = pd.to_datetime(
    df_clean["Ship Date"],
    errors="coerce"
)

print(df_clean[["Order Date", "Ship Date"]].dtypes)
display(df_clean[["Order Date", "Ship Date"]].head())
```

```python
df_clean["Shipping Days"] = (
    df_clean["Ship Date"] - df_clean["Order Date"]
).dt.days

df_clean["Year"] = df_clean["Order Date"].dt.year
df_clean["Month"] = df_clean["Order Date"].dt.month_name()

display(
    df_clean[
        ["Order Date", "Ship Date", "Shipping Days", "Year", "Month"]
    ].head()
)

print("New columns created successfully!")
```

```python
df_clean["Profit Margin (%)"] = (
    df_clean["Profit"] / df_clean["Sales"]
) * 100

print("Profit margin column created!")
display(
    df_clean[
        ["Sales", "Profit", "Profit Margin (%)"]
    ].head()
)
```

```python
total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders = df_clean["Order ID"].nunique()
total_quantity = df_clean["Quantity"].sum()

print("===== SUPERSTORE SALES SUMMARY =====")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Unique Orders: {total_orders:,}")
print(f"Total Quantity Sold: {total_quantity:,.0f}")
print(f"Overall Profit Margin: {(total_profit / total_sales) * 100:.2f}%")
```

```python
category_summary = df_clean.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).round(2)

category_summary = category_summary.sort_values(
    by="Total_Sales", ascending=False
)

display(category_summary)
```

```python
import matplotlib.pyplot as plt

category_sales = df_clean.groupby("Category")["Sales"].sum()

category_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Total Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_profit = df_clean.groupby("Category")["Profit"].sum()

category_profit.plot(kind="bar", figsize=(8, 5))

plt.title("Total Profit by Category")
plt.xlabel("Product Category")
plt.ylabel("Total Profit ($)")
plt.axhline(y=0, color="black", linewidth=0.8)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_sales = df_clean.groupby("Region")["Sales"].sum()
region_sales = region_sales.sort_values(ascending=False)

display(region_sales)

region_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
monthly_sales = df_clean.groupby(
    df_clean["Order Date"].dt.to_period("M")
)["Sales"].sum()

display(monthly_sales)
```

```python
monthly_sales.plot(
    kind="line",
    figsize=(12, 5),
    title="Monthly Sales Trend",
    ylabel="Sales ($)",
    xlabel="Month"
)

import matplotlib.pyplot as plt
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
monthly_profit = df_clean.groupby(
    df_clean["Order Date"].dt.to_period("M")
)["Profit"].sum()

monthly_profit.plot(
    kind="line",
    figsize=(12, 5),
    title="Monthly Profit Trend",
    ylabel="Profit ($)",
    xlabel="Month"
)

import matplotlib.pyplot as plt
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
